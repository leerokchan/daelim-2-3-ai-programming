```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

data = {
    "study_hours": [1, 2, 3, 3, 4, 4, 5, 6, 7, 8],
    "attendance": [50, 55, 60, 80, 60, 75, 75, 80, 90, 95],
    "pass_fail": [0, 0, 0, 1, 0, 1, 1, 1, 1, 1]     #딕셔너리파일이라고 부름, JAVA에서는 JSON
}

df = pd.DataFrame(data)

# 입력값과 정답 분리
X = df[["study_hours", "attendance"]]
y = df["pass_fail"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X,y)

print("데이터 준비 완료")
print(f"학생 수: {len(df)}명") # 학생 수: %%명
```

# 모델 정확도

학습에 사용한 데이터로 정확도 확인

```python
accuracy = model.score(X,y)
print("모델 정확도:", accuracy) # 0~1 표시됨 -> 0~100%
print(f"모델 정확도 퍼센트:{accuracy * 100}%")
```

```python
# 모델이 어떤 예측을 했는지 확인
predictions = model.predict(X)

# Columns - 공부시간, 출석률, 실제, 예측, 결과
# zip() → zip([1,2,3], [4,5,6]) → (1,4), (2,5), (3,6) ---zip함수에 대해서 개념요약 예시문
# 실제값과 예측값을 한쌍으로 묶어서...
result = pd.DataFrame({
    "공부시간": X["study_hours"].values,
    "출석률": X["attendance"].values,
    "실제": y.values, # x에 비해서 값이 한개이기때문에 배열식으로 입력해주지 않고, 바로 value를 찾는다
    "예측": predictions,
    "결과": ["O" if a == p else "X" for a. p in zip(y,predictions)]
        #for문과 반복문을 같이 쓸 때 파이썬에서는 참값을 먼저 적고, 그 조건을 적는다. 만약 참 값이 아니면 else 값을 출력 할 수 있도록 출력 값을 적고, for문을 가장 마지막에 입력시킨다.
})

result
```
